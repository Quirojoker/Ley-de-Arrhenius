##Ley de Arrhenius (Química)
---


En química, la velocidad de una reacción depende fuertemente de la temperatura. Esta relación se describe matemáticamente mediante la ley de Arrhenius, que establece que la constante de velocidad $k$ crece exponencialmente con la temperatura $T$, controlada por dos parámetros: la energía de activación $E_a$, que mide la barrera energética que las moléculas deben superar para reaccionar, y el factor preexponencial $A$, que representa la frecuencia de colisiones efectivas entre moléculas. Este modelo no solo explica por qué las reacciones ocurren más rápido al aumentar la temperatura, sino que además permite estimar $E_a$ y $A$ a partir de datos experimentales, proporcionando información fundamental para comprender y diseñar procesos químicos.


<br>

La ley de Arrhenius relaciona la constante de velocidad $k$ con la temperatura absoluta $T$ como:

$$
k(T) = A\,e^{-E_a/(R\,T)},
$$


donde $A$ es el factor preexponencial, $E_a$ la energía de activación y $R=8.314\ \mathrm{J\,mol^{-1}K^{-1}}$.


Suponga que se realiza un experimento y se obtienen los siguientes datos experimentales (temperatura $T$ en K, constante de velocidad $k$ en s $^{-1}$):

## Solución
## 1. Cambio de Variables

$$k = A\cdot \exp\left( -~\frac{E}{R\cdot T}\right),$$

$$ln(k) = ln[A\cdot \exp\left(-~\frac{E}{R\cdot T}\right)],$$

$$ln(k) = ln(A) + ln(\exp\left(-~\frac{E}{R\cdot T}\right)),$$

$$ln(k) = ln(A) + (-~\frac{E}{R\cdot T})\cdot ln(\exp),$$

$$ln(k) = ln(A) + (-~\frac{E}{R\cdot T})\cdot 1,$$

$$ln(k) = ln(A) + (-~\frac{E}{R\cdot T}),$$

$$ y = ln(k),$$

$$ x = (~\frac{1}{R\cdot T}),$$

$$ \exp(b) = (A),$$

$$ E = -m,$$

## como tal tendriamos la ecuación de la recta 

$$ y = ln(A) - (E\cdot x),$$

```python

import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import scipy.stats as stat 
from scipy.interpolate import lagrange

Temperatura_kelvin = [280, 
    285, 
    290, 
    295, 
    300, 
    305, 
    310, 
    315, 
    320, 
    330, 
    340]

K_constante_velocidad = [4.6995e2,
    6.8501e2,
    9.8560e2,
    1.4007e3,
    1.9675e3,
    2.7330e3,
    3.7562e3,
    5.1108e3,
    6.8872e3,
    1.2172e4,
    2.0805e4]

df = pd.DataFrame({'T(k)':Temperatura_kelvin, 'K(s^-1)':K_constante_velocidad})

print(f'Datos obtenidos en el Laboratorio: \n{df}')
print()

# Extraccion de datos y pasar a arreglos.

T_kelvin = np.array(df['T(k)'])
K_medido = np.array(df['K(s^-1)'])

print(f'Temperatura: \n{T_kelvin} k')
print(f'constante de velocidad de reacción: \n{K_medido} 1/s')
print()

# Grafica de datos originales (T(k), K(1/s))
print('Grafica 1. Datos obtenidos en el laboratorio')
plt.plot(Temperatura_kelvin, K_constante_velocidad, marker = 'd', label ='T vs K' )
plt.title('Gráfica de los datos originales (T vs K)')
plt.legend()
plt.xlabel('T(k)')
plt.ylabel('K 1/s')
plt.grid()
plt.show()
print()


# Linealización

# Parametro de la constante de los gases

R = 8.314 # unidades de J/(mol*K)

# Ecuacion de cambio de variables

datos_x = 1/(R*T_kelvin) # unidades de 1/(Jules/mol) o (mol/jules)
datos_y = np.log(K_medido) # unidades de (1/s) o (s^-1)

print('-'*10,'Cambio de variables', '-'*10)
print()
print(f'Datos de x: \n{datos_x} mol/jules')
print(f'Datos de y: \n{datos_y} s^-1')
print()


# Grafica Determinación de la prediccion del modelo 

print(f'Grafica 2.Determinación de la prediccion del modelo ')
plt.plot(datos_x, datos_y, marker = 'd', linestyle = '--', label = 'Modelo de Regresión Lineal')
plt.title('Regresión Lineal (Ln(K) vs K )')
plt.legend()
plt.xlabel('X = 1/(R*T)')
plt.ylabel('Y = Ln(K)')
plt.grid()
plt.show()
print()


# Regresión Lineal
#La regresión lineal siempre busca determinar la pendiente y el intercepto de la recta que mejor se ajusta a los datos.

regresion_lineal = stat.linregress(datos_x, datos_y)

# Parametros de la Regresion Lineal

intercepto = regresion_lineal.intercept
pendiente = regresion_lineal.slope

print('-'*10, 'Los Parametros de la Regresión Lineal son', '-'*10)
print(f'pendiente: {pendiente: .2e}')
print(f'intercepto: {intercepto: .2e}')
print()


# Parametros del modelo

A_mod = np.exp(intercepto)# A= e^b
E_mod_kilo_julios= -(pendiente/1000) # conversion de jules a kilojules
E_mod_julios= -(pendiente)

print('-'*10, 'Los Parámetros del Modelo Obtenidos son', '-'*10)
print(f'Coeficiente de frecuencia A = {A_mod: .2e} 1/s')
print(f'Energía de activación E en kilojulios = {E_mod_kilo_julios: .2e} KJ/mol')
print(f'Energía de activación E en julios = {E_mod_julios: .2e} J/mol')
print()

# Usando modelo para predecir

# utilizando la ecuacion de Arrhenius para predecir la constante de velocidad de reacción a T = 350 k
T_350 = 350
K_350 = A_mod*np.exp(-E_mod_julios/(R * T_350))
print(f'La velocidad de reacción utilizando la ecuación de Arrhenius para T = 350 k es K: {K_350: .2e} 1/s')
print()
# utilizando la ecuacion de la recta para predecir la constante de velocidad de reacción a T = 350 k
X_350_lineal = 1/(R*T_350)
ln_k_350 = pendiente*X_350_lineal + intercepto
K_350_lineal = np.exp(ln_k_350)
print(f'la velocidad de reacción utilizando la ecuación de la recta para T = 350 k es K: {K_350_lineal: .2e} 1/s')
print()

# Interpolación de lagrange

T_datos_x = np.array(df['T(k)'][:4]) #T_datos_x = np.array([280, 285, 290, 295])
K_datos_y = np.array(df['K(s^-1)'][:4])#K_datos_y = np.array([4.6995e2, 6.8501e2, 9.8560e2, 1.4007e3])

print('-'*10, 'Interpolación', '-'*10)
print(f'datos de temperatura en x: {T_datos_x}')
print(f'datos de constante de velocidad de reacción en y: {K_datos_y}')
print()

# Calculo de polinomio interpolante

polinomio_lagrange = lagrange(T_datos_x, K_datos_y)
print(f'Polinomio de Lagrange: \n{polinomio_lagrange}')
print()

x_plot = np.linspace(280, 295, 100)
y_plot = polinomio_lagrange(x_plot)

# Gráfica del polinomio de lagrange

print('Grafica 3. Polinomio de Lagrange')
plt.scatter(T_datos_x, K_datos_y, color='red', label='Datos de laboratorio')
plt.plot(x_plot, y_plot, label = 'Polinomio de Lagrange')
plt.title('Polinomio de Lagrange')
plt.legend()
plt.xlabel('T(k)')
plt.ylabel('K(1/s)')
plt.grid()
plt.show()

# Predicción utilizando el polinomio de lagrange y el modelo de arrhenius T= 288 K
print('-'*10, 'Predicción utilizando el polinomio de lagrange y el modelo de arrhenius T= 288 K', '-'*10)
print()

T_288 = 288
K_288_lagrange = polinomio_lagrange(T_288)
print(f'La predicción utilizando el polinomio de Lagrange para T = 288 k es K: {K_288_lagrange: .2e} 1/s')
print()

K_288_arrhenius = A_mod*np.exp(-E_mod_julios/(R * T_288))
print(f'La predicción utilizando el modelo de Arrhenius para T = 288 k es K: {K_288_arrhenius: .2e} 1/s')
print()
```
